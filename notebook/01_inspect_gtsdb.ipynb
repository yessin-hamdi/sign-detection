```python
GT_PATH = "../data/FullIJCNN2013/gt.txt"
with open(GT_PATH) as f:
    lines = f.readlines()

print(len(lines))
print(lines[:3])
```

```python
total_signs = 0
unique_images = set()
signs_per_image = {}
widths = []
for line in lines:
    line = line.strip()
    if not line:
        continue

    filename, x1, y1, x2, y2, class_id = line.split(";")
    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
    class_id = int(class_id)
    total_signs += 1
    unique_images.add(filename)
    signs_per_image[filename] = signs_per_image.get(filename, 0) + 1
    widths.append(x2 - x1)
```

```python
background_images = 900 - len(unique_images)
max_signs = max(signs_per_image.values())
average_width = sum(widths) / len(widths)

print(f"Total signs: {total_signs}")
print(f"Images with signs: {len(unique_images)}")
print(f"Background images: {background_images}")
print(f"Most signs in one image: {max_signs}")
print(f"Smallest width: {min(widths)} px")
print(f"Largest width: {max(widths)} px")
print(f"Average width: {average_width:.1f} px")
print(f"Signs narrower than 32 px: {sum(1 for w in widths if w < 32)}")
```

```python
assert sum(signs_per_image.values()) == total_signs, "Sign counts don't add up"
assert len(unique_images) == len(signs_per_image), "Set and dictionary disagree on image count"
assert min(widths) > 0, "Found a box with zero or negative width"

print("All sanity checks passed")
```
