```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

results = model.train(
    data="/content/yolo_dataset/data.yaml",
    imgsz=1280,
    epochs=100,
    patience=20,
    batch=-1,
    seed=42,
    project="/content/drive/MyDrive/sign-detection/runs",
    name="run_1280",
)
```

## Results (validation set, best.pt)

| Metric | 640 | 1280 |
|---|---|---|
| Precision | 0.965 | 0.994 |
| Recall | 0.902 | 0.977 |
| mAP50 | 0.962 | 0.993 |
| mAP50-95 | 0.808 | 0.874 |
| Batch size (AutoBatch) | 23 | 9 |
| Epochs run | 100 | 43 (early stop, best at 23) |
| Training time | ~1.0 h | ~0.7 h |
| Ultralytics version | 8.4.173 | 8.4.175 |

**Decision:** keep the 1280 model. The recall gain (+0.075) is more than twice the epoch-to-epoch noise (±0.03), and every metric improved.

**Notes:**
- Missed signs dropped from about 18 to about 4 out of 184, which supports the hypothesis that small signs were being lost at 640 (to be confirmed by analyzing the misses).
- The 1280 run stopped before epoch 90, so it never reached the `close_mosaic` phase that helped the 640 run.
- The two runs used different Ultralytics versions, so the version is now pinned.
- These are validation scores; the honest final number will come from the test set.
